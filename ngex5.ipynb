AIM #2: Finding issues (empty, NAs, incorrect value, incorrect format, outliers, etc.) 
1. Find out how many missing values there are in the dataset
2. For the 'Age' column, find the best way to handle the missing values
    2.1. Use an appropriate plot to study the nature of the 'Age' column
    2.2. Figure out what is the best way to calculate the central tendency of the 'Age' column based on the above plot
    2.3. Using the most suitable central tendency measure, fill the missing values in the age column
3. Decide what is the best way to handle the missing values in the 'Cabin' columns
4. Similarly, decide what is the best way to handle the missing values in the 'Embarked' columns
5. Handle the incorrect data under the 'Survived' columns using appropriate measure
6. Handle the incorrectly formatted data under the 'Fare' column


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
file_path = 'titanic.csv'
df = pd.read_csv(file_path)

# 1. Find out how many missing values there are in the dataset
missing_values = df.isnull().sum()
print("Missing Values:\n", missing_values)

# 2. Handling missing values in the 'Age' column
# 2.1. Use a plot to study the nature of the 'Age' column
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'].dropna(), kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# 2.2. Decide the best measure of central tendency based on the plot
# If the distribution is skewed, consider using the median
age_median = df['Age'].median()

# 2.3. Fill missing values in the 'Age' column using the median
df['Age'].fillna(age_median, inplace=True)

# 3. Handle missing values in the 'Cabin' column
# If most are missing, drop the column or fill with 'Unknown'
cabin_missing_ratio = df['Cabin'].isnull().mean()
if cabin_missing_ratio > 0.5:
    df.drop('Cabin', axis=1, inplace=True)
else:
    df['Cabin'].fillna('Unknown', inplace=True)

# 4. Handle missing values in the 'Embarked' column
# Fill with the mode (most frequent value)
embarked_mode = df['Embarked'].mode()[0]
df['Embarked'].fillna(embarked_mode, inplace=True)

# 5. Handle incorrect data under the 'Survived' column
# Ensure values are binary (0 or 1). If not, correct them.
df['Survived'] = df['Survived'].apply(lambda x: 1 if x == '1' else 0)

# 6. Handle incorrectly formatted data under the 'Fare' column
# Ensure all 'Fare' values are numeric
df['Fare'] = pd.to_numeric(df['Fare'], errors='coerce').fillna(0)

# Display the cleaned dataset
print("\nCleaned Dataset Preview:\n", df.head())
```

AIM #3: Grouping 
1. Find out the average fare grouped by Pclass
    1.1. Plot the above using a suitable plot
2. Find out the average fare grouped by Sex
    2.1. Plot the above using a suitable plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Titanic dataset
file_path = 'titanic.csv'  
df = pd.read_csv(file_path)

# Aim #3: Grouping

# 1. Find out the average fare grouped by Pclass
average_fare_pclass = df.groupby('Pclass')['Fare'].mean()
print("\nAverage Fare by Pclass:\n", average_fare_pclass)

# 1.1. Plot the average fare by Pclass
plt.figure(figsize=(8, 5))
sns.barplot(x=average_fare_pclass.index, y=average_fare_pclass.values, palette='viridis')
plt.title('Average Fare by Pclass')
plt.xlabel('Pclass')
plt.ylabel('Average Fare')
plt.xticks(rotation=0)  # Ensure x labels are horizontal
plt.grid(axis='y')
plt.show()

# 2. Find out the average fare grouped by Sex
average_fare_sex = df.groupby('Sex')['Fare'].mean()
print("\nAverage Fare by Sex:\n", average_fare_sex)

# 2.1. Plot the average fare by Sex
plt.figure(figsize=(8, 5))
sns.barplot(x=average_fare_sex.index, y=average_fare_sex.values, palette='plasma')
plt.title('Average Fare by Sex')
plt.xlabel('Sex')
plt.ylabel('Average Fare')
plt.xticks(rotation=0)  # Ensure x labels are horizontal
plt.grid(axis='y')
plt.show()
```

AIM #4: Dataset visualization using pandas

1. Plot the distribution of 'Age' using a suitable plot
2. Plot the distribution of 'Fare' using a suitable plot
3. Plot the distribution of 'Pclass' using a suitable plot
4. Plot the distribution of 'Survived' using a suitable plot
5. Plot the distribution of 'Embarked' using a suitable plot
6. Plot the distribution of 'Fare' grouped by 'Survived'
7. Plot the distribution of 'Fare' grouped by 'Pclass'
8. Plot the distribution of 'Age' grouped by 'Survived'
9. Plot the distribution of 'Age' grouped by 'PClass'
10. Combine the 'SibSp' and 'Parch' and plot its distribution grouped by 'Survived'

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Titanic dataset
file_path = 'titanic.csv'
df = pd.read_csv(file_path)

# Set the style for the plots
sns.set(style="whitegrid")

# Aim #4: Dataset Visualization

# 1. Plot the distribution of 'Age' using a suitable plot
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'].dropna(), bins=30, kde=True, color='blue')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()

# 2. Plot the distribution of 'Fare' using a suitable plot
plt.figure(figsize=(10, 6))
sns.histplot(df['Fare'].dropna(), bins=30, kde=True, color='orange')
plt.title('Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()

# 3. Plot the distribution of 'Pclass' using a suitable plot
plt.figure(figsize=(10, 6))
sns.countplot(x='Pclass', data=df, palette='pastel')
plt.title('Distribution of Pclass')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.grid(axis='y')
plt.show()

# 4. Plot the distribution of 'Survived' using a suitable plot
plt.figure(figsize=(10, 6))
sns.countplot(x='Survived', data=df, palette='pastel')
plt.title('Distribution of Survived')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.grid(axis='y')
plt.show()

# 5. Plot the distribution of 'Embarked' using a suitable plot
plt.figure(figsize=(10, 6))
sns.countplot(x='Embarked', data=df, palette='pastel')
plt.title('Distribution of Embarked')
plt.xlabel('Embarked Port')
plt.ylabel('Count')
plt.grid(axis='y')
plt.show()

# 6. Plot the distribution of 'Fare' grouped by 'Survived'
plt.figure(figsize=(10, 6))
sns.boxplot(x='Survived', y='Fare', data=df, palette='pastel')
plt.title('Fare Distribution by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Fare')
plt.grid(axis='y')
plt.show()

# 7. Plot the distribution of 'Fare' grouped by 'Pclass'
plt.figure(figsize=(10, 6))
sns.boxplot(x='Pclass', y='Fare', data=df, palette='pastel')
plt.title('Fare Distribution by Pclass')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.grid(axis='y')
plt.show()

# 8. Plot the distribution of 'Age' grouped by 'Survived'
plt.figure(figsize=(10, 6))
sns.boxplot(x='Survived', y='Age', data=df, palette='pastel')
plt.title('Age Distribution by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Age')
plt.grid(axis='y')
plt.show()

# 9. Plot the distribution of 'Age' grouped by 'Pclass'
plt.figure(figsize=(10, 6))
sns.boxplot(x='Pclass', y='Age', data=df, palette='pastel')
plt.title('Age Distribution by Pclass')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.grid(axis='y')
plt.show()

# 10. Combine 'SibSp' and 'Parch' and plot its distribution grouped by 'Survived'
df['FamilySize'] = df['SibSp'] + df['Parch']
plt.figure(figsize=(10, 6))
sns.countplot(x='FamilySize', hue='Survived', data=df, palette='pastel')
plt.title('Family Size Distribution by Survival')
plt.xlabel('Family Size (SibSp + Parch)')
plt.ylabel('Count')
plt.grid(axis='y')
plt.show()

# 11. Combine 'SibSp' and 'Parch' and plot its distribution grouped by 'Pclass'
plt.figure(figsize=(10, 6))
sns.countplot(x='FamilySize', hue='Pclass', data=df, palette='pastel')
plt.title('Family Size Distribution by Pclass')
plt.xlabel('Family Size (SibSp + Parch)')
plt.ylabel('Count')
plt.grid(axis='y')
plt.show()

# 12. Plot a distribution between 'Age' and 'Fare' to see if there's any relationship
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Age', y='Fare', data=df, alpha=0.6, color='green')
plt.title('Relationship between Age and Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid()
plt.show()

# 13. Other possibilities to show relationships
# You could use a pairplot to visualize relationships between multiple variables
sns.pairplot(df[['Age', 'Fare', 'Pclass', 'Survived']], hue='Survived', palette='pastel')
plt.suptitle('Pairplot of Selected Features', y=1.02)
plt.show()
```

AIM #5: Correlation

1. Generate a correlation matrix for the entire dataset
2. Find correlation between 'Age' and 'Fare'
3. What other possible correlations can be found in the dataset?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Titanic dataset
file_path = 'titanic.csv'
df = pd.read_csv(file_path)

# Aim #5: Correlation

# 1. Generate a correlation matrix for the entire dataset
correlation_matrix = df.corr()

# Plot the correlation matrix
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm', square=True, linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()

# 2. Find correlation between 'Age' and 'Fare'
age_fare_correlation = df['Age'].corr(df['Fare'])
print(f"\nCorrelation between Age and Fare: {age_fare_correlation:.2f}")

# 3. Other possible correlations in the dataset
# Displaying correlations greater than a threshold (e.g., 0.3 or -0.3)
threshold = 0.3
high_correlations = correlation_matrix[(correlation_matrix.abs() > threshold) & (correlation_matrix != 1)]

print("\nOther possible correlations in the dataset (greater than threshold):")
print(high_correlations.dropna(how='all', axis=0).dropna(how='all', axis=1))
```
